# 텍스트 변환기 - 여러 출력 한번에
- 하나의 입력으로 여러 개의 결과를 동시에 보여주는 법
- outputs에 리스트를 넘기면 화면에 출력창이 여러 개 생성

```python
import gradio as gr

def transform_text(sentence):
    upper_text = sentence.upper()          # 대문자로 변환
    lower_text = sentence.lower()          # 소문자로 변환
    reversed_text = sentence[::-1]         # 문자열 뒤집기 (슬라이싱 활용)
    length = len(sentence)                 # 문자열 길이

    # 함수가 값을 여러 개 반환하면, outputs 리스트 순서대로 매칭됩니다.
    return upper_text, lower_text, reversed_text, length
```

```python
demo = gr.Interface(
    fn=transform_text,
    # 입력창 크기 조절: lines=3, max_lines=5
    inputs=gr.Textbox(label="문장을 입력하세요", placeholder="예: Hello Gradio"),
    outputs=[
        gr.Textbox(label="대문자 변환"),
        gr.Textbox(label="소문자 변환"),
        gr.Textbox(label="문자열 뒤집기"),
        gr.Number(label="글자 수"),
    ],
    title="텍스트 변환기",
)
demo.launch()
```

```python
demo.close()
```
