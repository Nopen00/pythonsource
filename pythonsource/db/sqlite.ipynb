```python
import sqlite3

# isolation_level : 트랜잭션
# autocommit 중지
conn = sqlite3.connect("./test.db", isolation_level = None)

# sql 을 데이터 베이스에 전달하고 결과를 받아오는 객체
cursor = conn.cursor();
```

```python
# 테이블 생성
sql = """create table if not exists users(id interger primary key,
username text,
email text,
phone text,
website text,
regdate text
)
"""

cursor.execute(sql)
```
